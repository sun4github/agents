```python

from dotenv import load_dotenv

load_dotenv(override=True)
```

```python
import os
from openai import OpenAI
openai = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

messages=[{'role': 'user', 'content': 'Reinsurance is a type of insurance that is used to transfer the risk of a loss to another insurance company. As an reinsurance brokerage, pick a area in cedent and market services, where it makes sense to utilize Agentic AI to improve the service and efficiency?'}]

response = openai.chat.completions.create(
    model = "gpt-4o-mini",
    messages = messages
)

print(response.choices[0].message.content)



```
